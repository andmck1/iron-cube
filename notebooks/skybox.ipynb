```python
from PIL import Image
```

```python
img = Image.open("../assets/skybox/starmap.png").convert("RGB")
```

```python
threshold = 30
pixels = img.load()
```

```python
for y in range(img.height):
    for x in range(img.width):
        r, g, b = pixels[x, y]

        if r < threshold and g < threshold and b < threshold:
            pixels[x, y] = (0, 0, 0)
```

```python
img.save("../assets/skybox/processed.png")
```
