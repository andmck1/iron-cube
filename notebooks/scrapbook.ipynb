```python
import math
```

```python
n_instances = 3_250_000
disc_height = 100
disc_ro = 35_000
disc_ri = 25
```

```python
def calculate_density(ni, dh, dro, dri):
    return ni / (2.0 * math.pi * dh) * 1.0 / ((dro ** 2) - (dri ** 2))
```

```python
density = calculate_density(n_instances, disc_height, disc_ro, disc_ri)
density
```

```python
new_ni = 100_000
new_dro = ((new_ni / n_instances) * (disc_ro ** 2)) ** 0.5
new_rho = calculate_density(new_ni, disc_height, new_dro, disc_ri)
print(new_dro)
print(new_rho / density)
```
